This notebook demonstrates how Decision Tree classifiers can be used for binary classification problems.

## 1. Salary Classification

The objective is to predict whether a person's salary is greater than $100,000 based on categorical features such as company, job and degree.

```python
import pandas as pd 
```

```python
df = pd.read_csv("salaries.csv")
df.head()
```

```python
inputs=df.drop(["salary_more_then_100k"],axis='columns')
target=df["salary_more_then_100k"]
```

```python
inputs
target
```

### Encoding Categorical Variables

```python
from sklearn.preprocessing import LabelEncoder
```

```python
company_encoder = LabelEncoder()
job_encoder = LabelEncoder()
degree_encoder = LabelEncoder()
```

```python
inputs['company_n'] = company_encoder.fit_transform(inputs['company'])
inputs['job_n'] = job_encoder.fit_transform(inputs['job'])
inputs['degree_n'] = degree_encoder.fit_transform(inputs['degree'])
```

```python
inputs=inputs.drop(['company','job','degree'],axis='columns')
```

```python
inputs
```

### Training the Decision Tree

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

X_train, X_test, y_train, y_test = train_test_split(
    inputs,
    target,
    test_size=0.2,
    random_state=42
)

model = DecisionTreeClassifier(random_state=42)
model.fit(X_train, y_train)

accuracy = model.score(X_test, y_test)
print(f"Accuracy: {accuracy:.2%}")
```

## 2. Titanic Survival Classification

In this section, a Decision Tree classifier is used to predict whether a passenger survived the Titanic disaster.

```python
titanic=pd.read_csv('titanic.csv')
titanic.head()
```

```python
titanic.isnull().sum()
```

```python
target=titanic.Survived
```

```python
titanic=titanic.drop(['PassengerId','Name','Cabin','Ticket','Parch','SibSp','Survived','Embarked'],axis='columns')
```

```python
titanic.head()
```

```python
titanic.Sex = titanic.Sex.map({'male': 1, 'female': 2})
```

```python
titanic.head()
```

target.head()

```python
titanic.isnull().sum()
```

```python
mean_age = titanic.Age.mean()
titanic.Age = titanic.Age.fillna(mean_age)
```

```python
titanic.isnull().sum()
```

```python
titanic
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(titanic,target,test_size=0.2)
```

```python
from sklearn import tree
model = tree.DecisionTreeClassifier()
```

```python
model.fit(X_train,y_train)
```

```python
accuracy = model.score(X_test, y_test)
print(f"Accuracy: {accuracy:.2%}")
```
