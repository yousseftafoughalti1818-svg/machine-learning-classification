# Handwritten Digits Classification with Logistic Regression

## Objective
The objective of this project is to classify handwritten digits from 0 to 9 using Logistic Regression.

The dataset used is the Digits dataset provided by Scikit-learn.

## 1. Import Libraries and Load the Dataset

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
```

## 2. Explore the Dataset

```python
digits= load_digits()
```

```python
dir(digits)
digits.data[1700]
```

```python
print("Number of samples:", digits.data.shape[0])
print("Number of features:", digits.data.shape[1])
print("Classes:", digits.target_names)
```

```python
for i in range(5):
    plt.matshow(digits.images[i], cmap="gray")
    plt.title(f"Label: {digits.target[i]}")
    plt.show()
```

## 4. Split the Dataset into Training and Test Sets

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    digits.data,
    digits.target,
    test_size=0.2,
    random_state=42
)
```

```python
len(X_test)
```

```python
len(X_train)
```

## 5. Train the Logistic Regression Model

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
```

## 6. Make Predictions

```python
prediction = model.predict([digits.data[1200]])

print("Predicted digit:", prediction[0])
print("Actual digit:", digits.target[1200])
```

```python
model.predict(X_test)
```

## 7. Evaluate the Model

```python
accuracy = model.score(X_test, y_test)
print(f"Accuracy: {accuracy:.2%}")
```

```python
y_predicted=model.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_predicted)
cm
```

```python
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=digits.target_names
).plot()

plt.title("Confusion Matrix")
plt.show()
```

```python
## Conclusion

In this project, Logistic Regression was used to classify handwritten digits from the Scikit-learn Digits dataset.

The model was trained on 80% of the data and evaluated on the remaining 20%. The accuracy and confusion matrix were used to evaluate its classification performance.
```
