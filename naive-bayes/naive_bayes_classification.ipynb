# Naive Bayes Classification

## Objective

This notebook demonstrates the use of Naive Bayes algorithms for two classification problems:

1. Titanic survival prediction using Gaussian Naive Bayes.
2. Email spam detection using Multinomial Naive Bayes.

## 1. Titanic Survival Prediction

The objective is to predict whether a passenger survived the Titanic disaster using Gaussian Naive Bayes.

```python
import pandas as pd
```

```python
df=pd.read_csv('titanic.csv')
df.head()
```

```python
df.drop(['PassengerId','Name','SibSp','Parch','Ticket','Cabin','Embarked'],axis='columns',inplace=True)
df.head()
```

```python
inputs = df.drop('Survived',axis='columns')
target = df.Survived
```

```python
dummies = pd.get_dummies(inputs.Sex)
dummies.head(3)
```

```python
inputs = pd.concat([inputs,dummies],axis='columns')
inputs.head(3)
```

```python
inputs.drop(['Sex','male'],axis='columns',inplace=True)
inputs.head(3)
```

```python
inputs.columns[inputs.isna().any()]
```

```python
inputs.Age[:10]
```

```python
inputs.Age = inputs.Age.fillna(inputs.Age.mean())
inputs.head()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    inputs,
    target,
    test_size=0.3,
    random_state=42,
    stratify=target
)
```

```python
from sklearn.naive_bayes import GaussianNB

model = GaussianNB()
model.fit(X_train, y_train)

accuracy = model.score(X_test, y_test)
print(f"Accuracy: {accuracy:.2%}")
```

## 2. Email Spam Detection

The objective is to classify text messages as spam or not spam using Multinomial Naive Bayes.

CountVectorizer is used to transform text messages into numerical features that can be processed by the classifier.

```python
dff = pd.read_csv("spam.csv")
dff.head()
```

```python
dff['spam']=dff['Category'].apply(lambda x: 1 if x=='spam' else 0)
dff.head()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(dff.Message,dff.spam)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
# Transformer les messages en données numériques
v = CountVectorizer()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

X_train_count = v.fit_transform(X_train)
X_test_count = v.transform(X_test)

# Créer et entraîner le modèle
spam_model = MultinomialNB()
spam_model.fit(X_train_count, y_train)

# Évaluer le modèle
accuracy = spam_model.score(X_test_count, y_test)

print(f"Accuracy: {accuracy:.2%}")
```

```python
emails = [
    "Hey mohan, can we get together to watch football game tomorrow?",
    "Upto 20% discount on parking, exclusive offer just for you. Don't miss this reward!"
]

emails_count = v.transform(emails)

predictions = spam_model.predict(emails_count)

for email, prediction in zip(emails, predictions):
    label = "Spam" if prediction == 1 else "Not Spam"
    print(f"{label}: {email}")
```

```python
X_test_count = v.transform(X_test)
spam_model.score(X_test_count, y_test)
```
